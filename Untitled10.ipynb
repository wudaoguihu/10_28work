```python
import torch

def predict(x1, x2, w1, w2, b):
    return w1 * x1 + w2 * x2 + b
```

```python
x1 = torch.tensor([-2., -1., 0., 1., 2., 3.])
x2 = torch.tensor([1., 0., 2., -1., 3., 1.])
y = torch.tensor([-6., -1., -5., 6., -4., 4.])
print('x1:', x1, 'shape:', tuple(x1.shape))
print('x2:', x2, 'shape:', tuple(x2.shape))
print('y: ', y, 'shape:', tuple(y.shape))
```

```python
# 1. 创建三个参数，初始0，开启梯度记录
w1 = torch.tensor(0.0, requires_grad=True)
w2 = torch.tensor(0.0, requires_grad=True)
b  = torch.tensor(0.0, requires_grad=True)

# 2. 优化器，交给它三个参数
opt = torch.optim.SGD([w1, w2, b], lr=0.1)

epochs = 2000
for epoch in range(epochs):
    # 前向：批量预测，一次性传入x1,x2（都是shape(6,)）
    y_pred = predict(x1, x2, w1, w2, b)
    
    # 绝对损失：每个样本误差 → 求平均
    loss = torch.abs(y - y_pred).mean()

    # 梯度清零、反向传播、更新
    opt.zero_grad()
    loss.backward()
    opt.step()

    # 可选：每隔200轮打印看loss下降
    if epoch % 200 == 0:
        print(f"epoch={epoch:4d}, loss={loss.item():.4f}, w1={w1.item():.4f}, w2={w2.item():.4f}, b={b.item():.4f}")

# 训练结束打印最终结果
print("\n====训练完成====")
print("w1 =", w1.item())
print("w2 =", w2.item())
print("b  =", b.item())
print("最终平均损失：", loss.item())
```

```python
### 问答
1. 一次预测需要2个输入张量x1和x2。x1、x2的shape都是(6,)，维度大小等于样本数量6；x1[i]和x2[i]共同组成第i个样本的两个特征，一共6个样本。

2. 预测公式不变；参数列表仍然是w1,w2,b，没有改变；损失从单个样本的绝对误差，改成多个样本误差求平均得到平均损失。优化器的基础步骤没有改变：前向计算、求损失、清零梯度、反向传播、更新参数。

3. 训练有效判断依据：训练过程中平均损失持续下降，最后稳定在很小的数值；模型预测值和真实y接近。训练结束记录输出的w1、w2、b和最终loss。

4. 参数初始全为0时，初始预测值和真实标签之间存在误差，损失不为0，可以计算得到梯度；优化器依靠梯度更新参数，因此可以正常开始学习。
```
