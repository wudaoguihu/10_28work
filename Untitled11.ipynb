```python
import torch
import torch.nn.functional as F

torch.set_printoptions(precision=4, sci_mode=False)

def show(name, value):
    print(f'{name}: {value}, shape={tuple(value.shape)}, dtype={value.dtype}')
```

```python
def predict(x):
    # x 是一个 0 到 1 之间的浮点标量张量
    return torch.stack([1 - x, x])

x = torch.tensor(0.8)
y = torch.tensor([1., 0.])  # 真实情况：没有行人
y_pred = predict(x)

show('x', x)
show('y（one-hot 真实标签）', y)
show('y_pred（两个预测概率）', y_pred)
show('预测概率之和', y_pred.sum())
```

```python
scores = torch.log(y_pred)  # 从本例的概率构造等价 logits
label = y.argmax()        # [1, 0] → 类别编号 0，dtype 为 torch.int64

show('scores（传给损失函数的分数）', scores)
show('softmax(scores)（还原概率）', scores.softmax(dim=0))
show('label（真实类别编号）', label)

loss = F.cross_entropy(scores, label)
show('loss', loss)
```

```python
manual_loss = -torch.log(y_pred[label])
show('手算：真实类别概率的负对数', manual_loss)
print('与 cross_entropy 一致：', torch.allclose(loss, manual_loss))
```

```python
def compare(y, probabilities):
    label = y.argmax()
    # 仅为展示指定概率的交叉熵。log(0)=-inf，是这里刻意保留的边界情况。
    scores = torch.log(probabilities)
    loss = F.cross_entropy(scores, label)
    print(f'y={y.tolist()}, y_pred={probabilities.tolist()}')
    print(f'真实类别概率={probabilities[label].item():.4f}, loss={loss.item():.4f}')
    return loss

y = torch.tensor([1., 0.])
example_losses = []
for values in [[0.9, 0.1], [0.5, 0.5], [0.1, 0.9], [0., 1.], [1., 0.]]:
    example_losses.append(compare(y, torch.tensor(values)))
    print()
```

```python
y = torch.tensor([0., 1.])
y_pred = torch.tensor([0.1, 0.9])
compare(y, y_pred)
```

```python
import torch
import torch.nn.functional as F

y = torch.tensor([0., 1.])
y_pred = torch.tensor([0.8, 0.2])

# TODO: 打印 y 和 y_pred 的 shape
print("y shape:", tuple(y.shape))
print("y_pred shape:", tuple(y_pred.shape))

# TODO:由真实 y 得到类别编号 label
label = y.argmax()
print("label =", label.item())

# TODO：由 y_pred 构造等价 scores(logits)，再调用 F.cross_entropy
# logit = log(p) ，因为 softmax(logits)=p
scores = torch.log(y_pred)
loss = F.cross_entropy(scores, label)

# TODO: 打印 loss
print("loss =", loss.item())
```

1. 修改x后的预测与损失：
 
原来x=0.2的时候，模型给正确类别的预测概率比较高，交叉熵损失小；
改成x=0.8之后，模型给正确类别的预测概率变低了。
交叉熵损失 L=-\log(p_{正确类})，正确类概率越小，-\log(p) 就越大，所以损失变大。
 
你运行后记录： y_pred = [0.2000,0.8000]，loss = 1.6094379425048828
 
2. “有行人”实验的结果：
 
 y shape: (2,)，y_pred shape: (2,) ； label=1 ；loss≈1.6094。
计算验证：-log(0.2)≈1.6094
 
3. 两个极端例子的解释：
 
交叉熵只看真实标签那一维的预测概率：
 
- 真实标签[1,0]（类别0），预测概率[0,1]：正确类概率=0，-\log(0) \to +\infty，损失无穷大。

- 真实标签[1,0]，预测概率[1,0]：正确类概率=1，-log(1)=0，损失等于0。
 
4. 概率与原始分数应该怎样传入损失函数：
 
F.cross_entropy(torch.log(y_pred), label)  才是对的。
原因：
 F.cross_entropy 内部会自动做softmax+log，等价于 \text{log_softmax(logits)}。
如果我们手里已经是概率y_{pred}，想要直接用这个概率算交叉熵，需要传入 \log(y_{pred})；
如果直接把概率 y_pred 传给 cross_entropy ，它会再做一次softmax，结果就错了。
